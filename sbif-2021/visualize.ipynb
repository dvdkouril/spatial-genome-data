```python
import pyarrow as pa

file_path = "out/dros.3.arrow"
with pa.ipc.open_file(file_path) as arrow_file:
    table = arrow_file.read_all()

output_stream = pa.BufferOutputStream()
with pa.ipc.new_stream(output_stream, table.schema) as writer:
    writer.write_table(table)

table_bytes = output_stream.getvalue().to_pybytes()

vc = {
    "color": "lightgreen", 
    "scale": 0.01, 
    "links": True, 
    "mark": "sphere"
}
chromospyce.Widget(structure=table_bytes, viewconfig=vc)
```
